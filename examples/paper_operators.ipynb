**How the operators are built.** Every operator is a linear combination of operators of the bundled database,
`Operator_from_database(id)`, whose IDs are those of the operator catalogue
(see [`operator_catalogue.ipynb`](operator_catalogue.ipynb) or the `h4lat-catalogue` command).
The operators are combined with the arithmetic shown in [`operator_algebra.ipynb`](operator_algebra.ipynb).
For each paper, the code of `get_paper_operators` that builds its operators is shown below.
Changing these combinations (the normalisations, the components, or the catalogue operators themselves)
gives the operator basis used in other works.

The comments in that code give the index structure of each operator, up to its normalisation.
The first index is that of $\gamma_\mu$ for vector and axial operators, and the first two are those of
$\sigma_{\mu\nu}$ for tensor operators. The other indices are those of the derivatives.
$\{\ldots\}$ is the symmetric combination of the indices in it and $[\ldots]$ the antisymmetric one
(in `[1,{3],4}` the indices 3 and 4 are symmetrised first, then 1 and 3 are antisymmetrised).

Each operator is shown with its label (`op.id`), H(4) irrep, C-parity and trace condition.

````python
from IPython.display import Markdown, display

from h4lat import AVAILABLE_PAPERS, get_paper_operators


def show(ops, X=None, irrep=None):
    """Display the operators with Dirac structure X and H(4) irrep irrep (None: any)."""
    for op in ops:
        if (X is None or op.X == X) and (irrep is None or op.irrep == irrep):
            display(Markdown(f"**`{op.id}` &emsp; irrep: {op.irrep} &emsp; C-parity: {op.C} &emsp; Trace condition: {op.tr}**"))
            op.display()
            display(Markdown("&nbsp;"))
````

### M. Göckeler *et al.*, Phys. Rev. D 71, 114511 (2005)

*A lattice determination of moments of unpolarised nucleon structure functions using improved Wilson fermions*
&emsp; [DOI](https://doi.org/10.1103/PhysRevD.71.114511) · [arXiv:hep-ph/0410187](https://arxiv.org/abs/hep-ph/0410187) · [INSPIRE](https://inspirehep.net/literature?q=texkey:Gockeler:2004wp)

Vector operators for the moments $v_2$ (1 derivative, irreps (6,3) and (3,1)), $v_3$ (2 derivatives, irrep (8,1))
and $v_4$ (3 derivatives, irrep (2,1)) of the unpolarised nucleon structure functions, given in Eq. (31) of the paper.
The $v_3$ and $v_4$ operators are each the sum of the same combination of components of blocks 1 and 2 of their irrep:
they are completely symmetric, but they do not belong to a single block (their `block` is `None`).

The operators are built in `get_paper_operators` as follows:

```python
# Operators of Eq. (31) of the paper, for the moments v_2, v_3 and v_4 of the unpolarised structure
# functions (the paper uses the momenta p = 0 and p = (p_1, 0, 0))

# Vector - irrep (6,3), C=+1
opv2a = Operator_from_database(38)         # = {1,4}
# Vector - irrep (3,1), C=+1
opv2b = -1/3 * Operator_from_database(26)  # = 4,4 - 1/3 * (1,1 + 2,2 + 3,3)

# Third moment, v_3 (2 derivatives): Vector - irrep (8,1), C=-1
# (the same combination of components 1 and 5 of blocks 1 and 2: each part has mixed symmetry,
#  their sum is completely symmetric, so the operator belongs to no single block)
opv3 = -1/2 * (Operator_from_database(137) + 3/2 * Operator_from_database(141) + Operator_from_database(145) + 3/2 * Operator_from_database(149))  # = {4,4,1} - 1/2 * ({2,2,1} + {3,3,1})

# Fourth moment, v_4 (3 derivatives): Vector - irrep (2,1), C=+1
# (likewise, the same combination of components 1 and 2 of blocks 1 and 2)
opv4 = (-Operator_from_database(471) - 3/2 * Operator_from_database(472) - Operator_from_database(473) - 3/2 * Operator_from_database(474)) / 2  # = {1,1,4,4} + {2,2,3,3} - 1/2 * ({1,1,2,2} + {1,1,3,3} + {2,2,4,4} + {3,3,4,4})

# Set custom labels
opv2a.id = "v_2a"
opv2b.id = "v_2b"
```

### D. Dolgov *et al.* (LHPC and SESAM), Phys. Rev. D 66, 034506 (2002)

*Moments of nucleon light cone quark distributions calculated in full lattice QCD*
&emsp; [DOI](https://doi.org/10.1103/PhysRevD.66.034506) · [arXiv:hep-lat/0201021](https://arxiv.org/abs/hep-lat/0201021) · [INSPIRE](https://inspirehep.net/literature?q=texkey:LHPC:2002xzk)

Operators of Table II of the paper, for the moments of the quark density ($\langle x \rangle_q$, $\langle x^2 \rangle_q$, $\langle x^3 \rangle_q$: vector, 1 to 3 derivatives),
helicity ($\langle 1 \rangle_{\Delta q}$, $\langle x \rangle_{\Delta q}$, $\langle x^2 \rangle_{\Delta q}$: axial, 0 to 2 derivatives) and transversity ($\langle 1 \rangle_{\delta q}$, $\langle x \rangle_{\delta q}$: tensor, 0 and 1 derivative) distributions,
and for the twist-3 matrix elements $d_1$ and $d_2$ (axial, 1 and 2 derivatives).
In the labels `V2`, `A1`, `T2`, ... the number is $n + 1$ for the moment $\langle x^n \rangle$.

* The paper writes the axial operators with $\gamma_5\gamma_\mu = -\gamma_\mu\gamma_5$ and the tensor ones with $\gamma_5\sigma_{\mu\nu}$.
  With $\gamma_5 = \gamma_1\gamma_2\gamma_3\gamma_4$, $\gamma_5\sigma_{34} = -\sigma_{12}$ and $\gamma_5\sigma_{31} = -\sigma_{24}$:
  `T1` is $\sigma_{12} = -\gamma_5\sigma_{34}$ and `T2` is $\gamma_5 (\sigma_{34} D_1 + \sigma_{31} D_4)$.
* For $\langle x \rangle_{\Delta q}$ and $\langle x^2 \rangle_{\Delta q}$ the paper gives the irreps $6_3^-$ and $4_2^+$, those of the vector operators with the same indices;
  the axial operators are in the irreps (6,4) and (4,3) (like $a_2$ in Göckeler *et al.* (1996) below).
* The paper gives the opposite C-parities for `T1`, `T2`, `d1` and `d2`. Here the C-parity of all these operators is
  $\eta_X (-1)^{n_\mathrm{der}}$, with $\eta_X = -1$ for $\sigma_{\mu\nu}$ and $+1$ for $\gamma_\mu\gamma_5$:
  `d1` has the C-parity of `A2_a` and `A2_b`, which have the same Dirac structure and number of derivatives.
* `V3` and `V4` are each the sum of the same combination of components of blocks 1 and 2 of their irrep:
  they are completely symmetric, but they do not belong to a single block (their `block` is `None`).
* The paper uses the nucleon momentum $\vec P = 0$ or $\vec P = (p_1, 0, 0)$ and polarises the nucleon in the 3-direction,
  as the default polarisation projector (`gamma_pol_index=0`) does.

### M. Göckeler *et al.*, Phys. Rev. D 53, 2317 (1996)

*Polarized and unpolarized nucleon structure functions from lattice QCD*
&emsp; [DOI](https://doi.org/10.1103/PhysRevD.53.2317) · [arXiv:hep-lat/9508004](https://arxiv.org/abs/hep-lat/9508004) · [INSPIRE](https://inspirehep.net/literature?q=texkey:Gockeler:1995wg)

Operators of Table II of the paper: vector operators for the moments $v_2$ (1 derivative, irreps (6,3) and (3,1)),
$v_3$ (2 derivatives, irrep (8,1)) and $v_4$ (3 derivatives, irrep (2,1)) of the unpolarised structure functions,
and axial operators for the moments $a_0$ (no derivative, irrep (4,4)), $a_2$ (2 derivatives, irrep (4,3))
and $d_2$ (2 derivatives, irrep (8,1)) of the polarised ones.

* The irreps $\tau^{(k)}_l$ of the paper are the irreps (k,l) of h4lat, and the C-parities of the paper are those computed here.
* `v_3` and `v_4` (other components than those of Göckeler *et al.* (2005) above) are each the sum of the same combination
  of components of blocks 1 and 2 of their irrep: they are completely symmetric, but they do not belong to a single block
  (their `block` is `None`).
* The paper uses the momentum $\vec p = 0$ for $v_{2,b}$ and $a_0$, and $\vec p = (p_1, 0, 0)$ for the others.
  It polarises the nucleon in the 2-direction, with $\Gamma = \frac{1}{2}(1 + \gamma_4)\, i\gamma_5\gamma_2$.
  The axial operators therefore use the polarisation projector `gamma_pol_index=1`,
  $\Gamma_\mathrm{pol} = \frac{1}{2}(1 + \gamma_4)(1 - i\gamma_1\gamma_3)$, which polarises the nucleon in the same direction:
  with the default projector, polarised in the 3-direction, their kinematic factor `op.K` would vanish at the momenta of the paper.
