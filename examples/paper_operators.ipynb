# Operators used in the literature

`get_paper_operators` returns the H(4)-irreducible lattice operators used in published lattice QCD works.
The paper is selected by its INSPIRE texkey, `get_paper_operators(paper="Rodekamp:2023wpe")`,
or by its DOI, `get_paper_operators(doi="10.1103/PhysRevD.109.074508")`.
The DOI is accepted with or without a resolver prefix such as `https://doi.org/`.

Older works are collected at the end, in the section *Less recent works*.

**How the operators are built.** Every operator is a linear combination of operators of the bundled database,
`Operator_from_database(id)`, whose IDs are those of the operator catalogue
(see [`operator_catalogue.ipynb`](operator_catalogue.ipynb) or the `h4lat-catalogue` command).
The operators are combined with the arithmetic shown in [`operator_algebra.ipynb`](operator_algebra.ipynb).
For each paper, the code of `get_paper_operators` that builds its operators is shown below.
Changing these combinations (the normalisations, the components, or the catalogue operators themselves)
gives the operator basis used in other works.

The comments in that code give the index structure of each operator, up to its normalisation.
The first index is that of $\gamma_\mu$ for vector and axial operators, and the first two are those of
$\sigma_{\mu\nu}$ for tensor operators. The other indices are those of the derivatives.
$\{\ldots\}$ is the symmetric combination of the indices in it.

Each operator is shown with its label (`op.id`), H(4) irrep, C-parity and trace condition.

````python
from IPython.display import Markdown, display

from h4lat import AVAILABLE_PAPERS, get_paper_operators


def show(ops, X=None, irrep=None):
    """Display the operators with Dirac structure X and H(4) irrep irrep (None: any)."""
    for op in ops:
        if (X is None or op.X == X) and (irrep is None or op.irrep == irrep):
            display(Markdown(f"**`{op.id}` &emsp; irrep: {op.irrep} &emsp; C-parity: {op.C} &emsp; Trace condition: {op.tr}**"))
            op.display()
            display(Markdown("&nbsp;"))
````

## S. Bürger *et al.*, Phys. Rev. D 105, 054504 (2022)

*Lattice results for the longitudinal spin structure and color forces on quarks in a nucleon*
&emsp; [DOI](https://doi.org/10.1103/PhysRevD.105.054504) · [arXiv:2111.08306](https://arxiv.org/abs/2111.08306) · [INSPIRE](https://inspirehep.net/literature?q=texkey:Burger:2021knd)

Operators for the moments $a_2$ (axial, irrep (4,3), 2 derivatives) and $d_2$ (axial, irrep (8,1), 2 derivatives,
together with the tensor operators of irrep (8,1) with 1 derivative).

The operators are built in `get_paper_operators` as follows:

```python

op3 = Operator_from_database(183)  # = {1,2,4}
op4 = Operator_from_database(184)  # = {1,2,3}

# d2: Axial - irrep (8,1)

op6 = 1/(4 * np.sqrt(3)) * (1/2 * Operator_from_database(208) + 3/2 * Operator_from_database(204))  # = 2x 2,{1,3} - 1,{2,3} - 3,{1,2}
op7 = 1/(4 * np.sqrt(3)) * (Operator_from_database(206) - 3/2 * Operator_from_database(202))        # = 2x 3,{1,4} - 1,{3,4} - 4,{1,3}
op8 = 1/(4 * np.sqrt(3)) * (Operator_from_database(205) + 3/2 * Operator_from_database(201))        # = 2x 3,{2,4} - 2,{3,4} - 4,{2,3}
op9 = 1/4 * (Operator_from_database(205) - 1/2 * Operator_from_database(201))                        # = 4,{2,3} - 2,{3,4}
op10 = 1/4 * (Operator_from_database(206) + 1/2 * Operator_from_database(202))                       # = 4,{1,3} - 1,{3,4}
op11 = 1/4 * (-1/2 * Operator_from_database(207) - 1/2 * Operator_from_database(203))                # = 4,{1,2} - 1,{2,4}
op12 = 1/4 * (-1/2 * Operator_from_database(208) + 1/2 * Operator_from_database(204))                # = 3,{1,2} - 1,{2,3}

# d2: Tensor - irrep (8,1)
op13 = 1/(2 * np.sqrt(2)) * (-1/2 * Operator_from_database(67) + 1/2 * Operator_from_database(71))  # = 3,1,1 - 3,4,4
op14 = 1/(2 * np.sqrt(2)) * (1/2 * Operator_from_database(68) + 1/2 * Operator_from_database(72))   # = 4,3,3 - 4,1,1

op18 = 1/(2 * np.sqrt(6)) * (Operator_from_database(66) + 3/2 * Operator_from_database(70))         # = -2x 2,3,3 + 2,1,1 + 2,4,4
op19 = 1/(2 * np.sqrt(6)) * (-1/2 * Operator_from_database(67) - 3/2 * Operator_from_database(71))  # = 2x 3,2,2 - 3,1,1 - 3,4,4
op20 = 1/(2 * np.sqrt(6)) * (1/2 * Operator_from_database(68) - 3/2 * Operator_from_database(72))   # = -2x 4,2,2 + 4,1,1 + 4,3,3

# Set custom labels
op1.id = "O^5_{234}"
op2.id = "O^5_{134}"
op3.id = "O^5_{124}"
op4.id = "O^5_{123}"
op5.id = "O^1_1"
op6.id = "O^2_1"
```

## C. Alexandrou *et al.*, arXiv:2607.03458 (2026)

*The nucleon unpolarized generalized form factors and Mellin moments up to fourth order*
&emsp; [DOI](https://doi.org/10.48550/arXiv.2607.03458) · [arXiv:2607.03458](https://arxiv.org/abs/2607.03458) · [INSPIRE](https://inspirehep.net/literature?q=texkey:Alexandrou:2026tjs)

Vector operators with 2 derivatives (irrep (4,2)) and 3 derivatives (irreps (1,2) and (6,2)).
Each (6,2) operator is the sum of the same component of the four (6,2) blocks of the catalogue: it is completely symmetric,
but it does not belong to a single block (its `block` is `None`).

The operators are built in `get_paper_operators` as follows:

```python
# 2 derivatives: Vector - irrep (4,2)
op1 = Operator_from_database(129)  # = {2,3,4}
op2 = Operator_from_database(130)  # = {1,3,4}
op3 = Operator_from_database(131)  # = {1,2,4}
op4 = Operator_from_database(132)  # = {1,2,3}

# 3 derivatives: Vector
# irrep (1,2)
op5 = Operator_from_database(469)  # = {1,2,3,4}
# irrep (6,2): the sum of the same component of blocks 1-4 is completely symmetric;
# only the three components in which the index 4 appears twice are used
op6 = -Operator_from_database(592) - Operator_from_database(598) - Operator_from_database(604) - Operator_from_database(610)  # = {4,4,2,3} - {1,1,2,3}
op7 = -Operator_from_database(593) - Operator_from_database(599) - Operator_from_database(605) - Operator_from_database(611)  # = {4,4,1,3} - {2,2,1,3}
op8 = -Operator_from_database(594) - Operator_from_database(600) - Operator_from_database(606) - Operator_from_database(612)  # = {4,4,1,2} - {3,3,1,2}

# Set custom labels

op2.id = "{1,3,4}"
op3.id = "{1,2,4}"
op4.id = "{1,2,3}"
op5.id = "{1,2,3,4}"
op6.id = "{4,4,2,3} - {1,1,2,3}"
op7.id = "{4,4,1,3} - {2,2,1,3}"
op8.id = "{4,4,1,2} - {3,3,1,2}"
```

### M. Göckeler *et al.*, Phys. Rev. D 71, 114511 (2005)

*A lattice determination of moments of unpolarised nucleon structure functions using improved Wilson fermions*
&emsp; [DOI](https://doi.org/10.1103/PhysRevD.71.114511) · [arXiv:hep-ph/0410187](https://arxiv.org/abs/hep-ph/0410187) · [INSPIRE](https://inspirehep.net/literature?q=texkey:Gockeler:2004wp)

Vector operators for the moments $v_2$ (1 derivative, irreps (6,3) and (3,1)), $v_3$ (2 derivatives, irrep (8,1))
and $v_4$ (3 derivatives, irrep (2,1)) of the unpolarised nucleon structure functions, given in Eq. (31) of the paper.
The $v_3$ and $v_4$ operators are each the sum of the same combination of components of blocks 1 and 2 of their irrep:
they are completely symmetric, but they do not belong to a single block (their `block` is `None`).

The operators are built in `get_paper_operators` as follows:

```python
# Operators of Eq. (31) of the paper, for the moments v_2, v_3 and v_4 of the unpolarised structure
# functions (the paper uses the momenta p = 0 and p = (p_1, 0, 0))

# Second moment, v_2 (1 derivative)
# Vector - irrep (6,3), C=+1
opv2a = Operator_from_database(38)         # = {1,4}
# Vector - irrep (3,1), C=+1
opv2b = -1/3 * Operator_from_database(26)  # = 4,4 - 1/3 * (1,1 + 2,2 + 3,3)

# Third moment, v_3 (2 derivatives): Vector - irrep (8,1), C=-1
# (the same combination of components 1 and 5 of blocks 1 and 2: each part has mixed symmetry,
#  their sum is completely symmetric, so the operator belongs to no single block)
opv3 = -1/2 * (Operator_from_database(137) + 3/2 * Operator_from_database(141) + Operator_from_database(145) + 3/2 * Operator_from_database(149))  # = {4,4,1} - 1/2 * ({2,2,1} + {3,3,1})

# Fourth moment, v_4 (3 derivatives): Vector - irrep (2,1), C=+1
# (likewise, the same combination of components 1 and 2 of blocks 1 and 2)
opv4 = (-Operator_from_database(471) - 3/2 * Operator_from_database(472) - Operator_from_database(473) - 3/2 * Operator_from_database(474)) / 2  # = {1,1,4,4} + {2,2,3,3} - 1/2 * ({1,1,2,2} + {1,1,3,3} + {2,2,4,4} + {3,3,4,4})

# Set custom labels
opv2a.id = "v_2a"
opv2b.id = "v_2b"
opv3.id = "v_3"
opv4.id = "v_4"

return [opv2a, opv2b, opv3, opv4]
```
