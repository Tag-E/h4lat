# Operator catalogue

A listing of the **most relevant operators** of the bundled database: the vector (V), axial (A)
and tensor (T) operators with one and two derivatives (IDs 25–80 and 113–336), organised by
number of derivatives and Dirac structure, in the order of the database, then by H(4) irrep.

**This notebook shows only a subset of the catalogue.**  The bundled database holds 1360
operators: also the scalar (S) and pseudoscalar (P) operators, the operators without
derivatives and those with three derivatives (see the table of IDs in the README).  All of
them can be seen in the PDF catalogue, written from the shell with the `h4lat-catalogue`
command (`-j` computes the kinematic factors on all cores)
```
h4lat-catalogue                       # S, P, V, A, T with 0, 1, 2 derivatives (IDs 1–336)
h4lat-catalogue --nder 0 1 2 3 -j     # the whole database (IDs 1–1360)
h4lat-catalogue --keys S0 P0 V3 -j    # only some groups
```
or from Python with `operator_show`, which takes the same selection
(`operator_show()`, `operator_show(nder=[0, 1, 2, 3])`, `operator_show(keys=["S0", "P0"])`).
Any other group can also be listed here with `show_group` of the cells below, e.g.
`show_group('S', 0)`.

Each group is loaded with `get_OperatorDict(X=..., nder=...)`, which returns the nested dict
```
d[(n, X)][(irrep, block)]  →  list[Operator]
```
where `n` is the number of Lorentz indices (`nder` for S and P, `nder + 1` for V and A,
`nder + 2` for T), `X` is the Dirac structure (`'S'`, `'P'`, `'V'`, `'A'` or `'T'`),
`irrep` is the H(4) irrep label `(k, l)`, and `block` is the multiplicity index.

For each operator we display:
* **id** — integer identifier in the bundled database
* **Operator form** — $\sum_\mu c_\mu\, O^X_\mu$ rendered as LaTeX via `op.display()`
* **Kinematic factor** — tree-level K via `op.K`


````python
from IPython.display import display, Markdown
from h4lat import get_OperatorDict, get_OperatorList


def show_group(X, nder):
    """Display every operator with Dirac structure X and nder derivatives, irrep by irrep."""
    (group,) = get_OperatorDict(X=X, nder=nder).values()  # {(irrep, block): [operators]}
    for (irrep, block), ops in group.items():
        display(Markdown(f"### Irrep **{irrep}**, block {block} &nbsp; ({len(ops)} operator{'s' if len(ops) > 1 else ''})"))
        for op in ops:
            display(Markdown(f"**id = {op.id}** &emsp; C-parity: {op.C} &emsp; Symmetry: {op.symm}  &emsp; Trace condition: {op.tr}"))
            print("Operator form:")
            op.display()
            print("Kinematic factor:")
            display(op.K)
            display(Markdown("&nbsp;"))


# The subset shown in this notebook: V, A and T with one and two derivatives
shown = get_OperatorList(X="VAT", nder=[1, 2])
print(f"{len(shown)} of the {len(get_OperatorList())} operators of the database are shown in this notebook.")
````

---
## Vector operators (Γ = γ_μ) — 1 derivative &nbsp;&nbsp;`(n=2, X='V')`

````python
show_group('V', 1)
````
